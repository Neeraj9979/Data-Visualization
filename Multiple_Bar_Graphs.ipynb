```python
import matplotlib.pyplot as plt

oscar_movies = [
    "The Dark Knight",
    "The Hurt Locker",
    "The King's Speech",
    "The Artist",
    "Argo",
]

years = [2008, 2009, 2010, 2011, 2012]
oscar_revenue = [1005, 170, 427, 133, 232]  # in $M
non_oscar_revenue = [378, 2788, 829, 185, 275]

# 1. Create a simple list of indices [0, 1, 2, 3, 4] using range()
x = list(range(len(years)))
width = 0.4

# 2. Plot the first bar set directly at the base positions
plt.bar(x, oscar_revenue, width, label="Oscar - Movies")

# 3. Shift the second bar set to the right using a list comprehension
plt.bar([i + width for i in x], non_oscar_revenue, width, label="Non Oscar - Movies")

# 4. Fix the X-axis ticks so they sit right in the middle of the grouped bars
plt.xticks([i + width/2 for i in x], years)

plt.title("Oscar Movies Revenue VS Non Oscar Movies Revenue in Each Year")
plt.xlabel("Years")
plt.ylabel("Revenue (in $M)")
plt.legend()

plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd

years = [2008, 2009, 2010, 2011, 2012]
oscar_revenue = [1005, 170, 427, 133, 232]  # in $M
non_oscar_revenue = [378, 2788, 829, 185, 275]

# 1. Combine your data into a dictionary
data = {
    "Oscar - Movies": oscar_revenue,
    "Non Oscar - Movies": non_oscar_revenue,
}

# 2. Create a DataFrame and use the years as the index
df = pd.DataFrame(data, index=years)

# 3. Plot a grouped bar chart directly from the DataFrame
df.plot(kind="bar", width=0.8)

# 4. Clean up the aesthetics using Matplotlib functions
plt.title("Oscar Movies Revenue VS Non Oscar Movies Revenue in Each Year")
plt.xlabel("Years")
plt.ylabel("Revenue (in $M)")
plt.xticks(rotation=0)  # Keeps the year labels horizontal
plt.legend()

plt.show()
```
