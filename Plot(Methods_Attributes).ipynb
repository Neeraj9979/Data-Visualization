```python
import matplotlib.pyplot as plt
oscar_movies = [
    "The Dark Knight",
    "The Hurt Locker",
    "The King's Speech",
    "The Artist",
    "Argo",
]

years = [2008, 2009, 2010, 2011, 2012]
oscar_revenue = [1005, 170, 427, 133, 232]  # in $M
non_oscar_movies = ["Slumdog Millionaire", "Avatar", "Inception", "Hugo", "Lincoln"]
non_oscar_revenue = [378, 2788, 829, 185, 275]
plt.plot(years,oscar_revenue,"+:c",label="Oscar - Movies")
plt.plot(years,non_oscar_revenue,color="#CBC3E3",marker="*",linestyle="-",linewidth=1,label="Non Oscar - Movies")
plt.title("Oscar Movies Revenue VS Non Oscar Movies Revenue in Each Year")
plt.xlabel("Years")
plt.ylabel("Revenue (in $M)")
plt.legend()

# plt.style.available
# plt.style.use('dark_background')
plt.grid()
plt.show()
```
