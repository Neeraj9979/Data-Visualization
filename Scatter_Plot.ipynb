```python
import matplotlib.pyplot as plt
# Scatter Plots
people = ["Person A", "Person B", "Person C", "Person D", "Person E",
          "Person F", "Person G", "Person H", "Person I", "Person J"]

age = [22, 25, 30, 35, 40, 45, 50, 55, 60, 65]
bp = [110, 115, 120, 122, 125, 130, 135, 123, 145, 150]

colours=["green" if x <120 else "red" for x in bp]

plt.scatter(age,bp,marker="*",color=colours,s=bp,alpha=0.5)
plt.title("Age Vs BP")
plt.xlabel("Age")
plt.ylabel("BP")
plt.grid()
```

```python
import matplotlib.pyplot as plt
# Scatter Plots
people = ["Person A", "Person B", "Person C", "Person D", "Person E",
          "Person F", "Person G", "Person H", "Person I", "Person J"]

age = [22, 25, 30, 35, 40, 45, 50, 55, 60, 65]
bp = [110, 115, 120, 122, 125, 130, 135, 123, 145, 150]

colours=["green" if x <120 else "red" for x in bp]

plt.scatter(age,bp,s=bp,cmap="viridis",c=bp)
plt.title("Age Vs BP")
plt.xlabel("Age")
plt.ylabel("BP")
plt.grid()
plt.colorbar(label="BP")

for i in range(len(bp)):
    plt.annotate(people[i],xy=(age[i],bp[i]), xytext=(age[i],bp[i]))

plt.xlim(20,74)
```
